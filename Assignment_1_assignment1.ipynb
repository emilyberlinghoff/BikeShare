# Part A: Search Problems

1. Node expansion shown in blue and red below.
- DFS (LIFO): $S→A→C→D→H→G$
- BFS (FIFO): $S→A→B→F→C→E→G$
- UCS (lowest path cost): $S→A→B→E→C→I→G$

2. Solution path shown in blue below.
- DFS (LIFO):
  - $S→A→C→D→H→H$
  - $1+1+1+1+1=5$
- BFS (FIFO):
  - $S→F→G$
  - $3+3=6$
- UCS (lowest path cost):
  - $S→B→E→I→G$
  - $1+0.5+0.5+0.5=2.5$

3. The discreprencies arise from different search strategies. DFS dives down one path without considerng cost or dopth, so it follows the alphabetical top path and finds a non-optimal but valid solution quickly. BFS guarentees the shallowest number of edges, not the lowest total cost, so it chooses $S→F→G$ (2 edges, total cost 6). UCS alwaays expands the lowest cumulative cost, so it prioritizes the $B→E→I$ edges and finds the true optimal path with total cost 2.5. BFS and DFS are uninformed while UCS uses cost information to guarentee optimality.
